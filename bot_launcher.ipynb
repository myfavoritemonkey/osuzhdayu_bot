```python
from configparser import ConfigParser
from pathlib import Path

config_path = Path('config.cfg')
if not config_path.exists():
    raise FileNotFoundError('Не найден config.cfg. Заполните его по образцу config.cfg.example.')

parser = ConfigParser()
parser.read(config_path, encoding='utf-8')
print('Конфигурация найдена.')
print('DeepSeek model:', parser.get('deepseek', 'model', fallback='не задана'))
print('Режим пересылки из ЛС:', parser.getboolean('telegram', 'relay_forwarded_messages', fallback=False))
# Секреты намеренно не выводятся.
```
